```python
#from langchain_community.llms import Ollama
from langchain_ollama import OllamaLLM

llm = OllamaLLM(
    #model="tinyllama:latest",
    model="qwen2.5:3b",
    temperature=0.2,
    num_ctx=2048,
    )

print(llm.invoke("¿Qué es son los LLMs(large language models) en el área de la IA?"))
```

```python
print(llm.invoke("¿Qué es son los LLMs en el área de la IA?"))
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in Artificial Intelligence for a technical audience."
)
response = llm.invoke(prompt.format(topic="LLMs"))
print(response)
```

```python
# En LangChain 1.2.6, LLMChain está deprecado. Usar directamente el LLM con el prompt
chain = prompt | llm

resultado = chain.invoke({"topic": "bias-variance tradeoff"})
print(resultado)
```
